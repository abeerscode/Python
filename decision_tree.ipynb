```python
import math
from collections import Counter
```

# ----------------------------------------------------------------------
# 1. TOY DATASET  (classic "Play Tennis" example — easy to hand-verify)
# ----------------------------------------------------------------------
# Each row: (Outlook, Temperature, Humidity, Wind) -> PlayTennis (label)

```python
data = [
    ("Sunny", "Hot", "High", "Weak", "No"),
    ("Sunny", "Hot", "High", "Strong", "No"),
    ("Overcast", "Hot", "High", "Weak", "Yes"),
    ("Rain", "Mild", "High", "Weak", "Yes"),
    ("Rain", "Cool", "Normal", "Weak", "Yes"),
    ("Rain", "Cool", "Normal", "Strong", "No"),
    ("Overcast", "Cool", "Normal", "Strong", "Yes"),
    ("Sunny", "Mild", "High", "Weak", "No"),
    ("Sunny", "Cool", "Normal", "Weak", "Yes"),
    ("Rain", "Mild", "Normal", "Weak", "Yes"),
    ("Sunny", "Mild", "Normal", "Strong", "Yes"),
    ("Overcast", "Mild", "High", "Strong", "Yes"),
    ("Overcast", "Hot", "Normal", "Weak", "Yes"),
    ("Rain", "Mild", "High", "Strong", "No"),
]

feature_names = ["Outlook", "Temperature", "Humidity", "Wind"]
```

# ----------------------------------------------------------------------
# Step-1 : Calculate Entrophy of full dataset
# ----------------------------------------------------------------------

```python
row = len(data)
print("Number of rows:", row)

yes_count = 0
no_count = 0

for row in data:
    if row[-1] == "Yes":
        yes_count += 1
    else:
        no_count += 1 

print("Yes count:", yes_count)
print("No count:", no_count)

probability_yes = yes_count / (yes_count + no_count)
probability_no = no_count / (yes_count + no_count)

print(f"Probability of Yes: {probability_yes:.3f}")
print(f"Probability of No: {probability_no:.3f}")

entrophy = - probability_yes * math.log2(probability_yes) - probability_no * math.log2(probability_no)
print(f"Entropy: {entrophy:.3f}")
```

# ----------------------------------------------------------------------
# Step-2 : Calculate Entrophy of Test Feature-1 (Outlook)
# ----------------------------------------------------------------------

```python
print("Split Sunny, Overcast, Rain by their labels (Yes/No)")

print()

sunny_yes_count = 0
sunny_no_count = 0

overcast_yes_count = 0
overcast_no_count = 0

rain_yes_count = 0
rain_no_count = 0


# Count Yes/No for Sunny, Overcast and Rain

for row in data:

    if row[0] == "Sunny" and row[-1] == "Yes":
        sunny_yes_count += 1

    if row[0] == "Sunny" and row[-1] == "No":
        sunny_no_count += 1

    if row[0] == "Overcast" and row[-1] == "Yes":
        overcast_yes_count += 1

    if row[0] == "Overcast" and row[-1] == "No":
        overcast_no_count += 1

    if row[0] == "Rain" and row[-1] == "Yes":
        rain_yes_count += 1

    if row[0] == "Rain" and row[-1] == "No":
        rain_no_count += 1


print("For Sunny: Yes =", sunny_yes_count, ", No =", sunny_no_count)

print("For Overcast: Yes =", overcast_yes_count, ", No =", overcast_no_count)

print("For Rain: Yes =", rain_yes_count, ", No =", rain_no_count)

print()


# --------------------------------------------------
# Entropy for Sunny
# --------------------------------------------------

print("Entropy for Sunny")

sunny_probability_yes = sunny_yes_count / (sunny_yes_count + sunny_no_count)

sunny_probability_no = sunny_no_count / (sunny_yes_count + sunny_no_count)

print(f"Probability of Sunny-Yes: {sunny_probability_yes:.3f}")

print(f"Probability of Sunny-No: {sunny_probability_no:.3f}")

sunny_entropy = -sunny_probability_yes * math.log2(sunny_probability_yes) - sunny_probability_no * math.log2(sunny_probability_no)

print(f"Entropy: {sunny_entropy:.3f}")

print()


# --------------------------------------------------
# Entropy for Overcast
# --------------------------------------------------

print("Entropy for Overcast")

overcast_probability_yes = overcast_yes_count / (overcast_yes_count + overcast_no_count)

overcast_probability_no = overcast_no_count / (overcast_yes_count + overcast_no_count)

print(f"Probability of Overcast-Yes: {overcast_probability_yes:.3f}")

print(f"Probability of Overcast-No: {overcast_probability_no:.3f}")


# Handle probability = 0
if overcast_probability_yes == 0:
    overcast_entropy_yes = 0
else:
    overcast_entropy_yes = -overcast_probability_yes * math.log2(overcast_probability_yes)


if overcast_probability_no == 0:
    overcast_entropy_no = 0
else:
    overcast_entropy_no = -overcast_probability_no * math.log2(overcast_probability_no)


overcast_entropy = overcast_entropy_yes + overcast_entropy_no

print(f"Entropy: {overcast_entropy:.3f}")

print()


# --------------------------------------------------
# Entropy for Rain
# --------------------------------------------------

print("Entropy for Rain")

rain_probability_yes = rain_yes_count / (rain_yes_count + rain_no_count)

rain_probability_no = rain_no_count / (rain_yes_count + rain_no_count)

print(f"Probability of Rain-Yes: {rain_probability_yes:.3f}")

print(f"Probability of Rain-No: {rain_probability_no:.3f}")


# Handle probability = 0
if rain_probability_yes == 0:
    rain_entropy_yes = 0
else:
    rain_entropy_yes = -rain_probability_yes * math.log2(rain_probability_yes)


if rain_probability_no == 0:
    rain_entropy_no = 0
else:
    rain_entropy_no = -rain_probability_no * math.log2(rain_probability_no)


rain_entropy = rain_entropy_yes + rain_entropy_no

print(f"Entropy: {rain_entropy:.3f}")
```
